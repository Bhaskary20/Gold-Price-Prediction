```python
import quandl
import numpy as np
import sklearn.linear_model
from sklearn.linear_model import LinearRegression
from sklearn.svm import SVR
from sklearn.model_selection import train_test_split
import yfinance as yf

```

```python
import pandas as pd

gold_df = yf.download("GC=F", start="2022-01-01")
print(gold_df.head())
```

```python
print(gold_df.columns.tolist())
```

```python
gold_df=gold_df[['Close']]
print(gold_df.head())
gold_df.columns = [col[0] for col in gold_df.columns]
```

```python
forecast_out=30
gold_df['prediction']=gold_df['Close'].shift(-forecast_out)
print(gold_df.head())
```

```python
print(gold_df.tail())
```

```python
x=np.array(gold_df.drop(['prediction'], axis=1))
x=x[:-forecast_out]
print(x)
```

```python
y=np.array(gold_df['prediction'])
y=y[: -forecast_out]
print(y)
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y, test_size=0.2)
```

```python
svr_rbf=SVR(kernel='rbf', C=1e3, gamma=0.1)
svr_rbf.fit(x_train, y_train)
```

```python
svm_confidence=svr_rbf.score(x_test, y_test)
print('svm_confidence: ', svm_confidence)
```

```python
lr=LinearRegression()
lr.fit(x_train, y_train)

lr_confidence=lr.score(x_test, y_test)
print('lr_confidence: ', lr_confidence)
```

```python
x_forecast=np.array(gold_df.drop(['prediction'], axis=1))[-forecast_out: ]

lr_prediction=lr.predict(x_forecast)
print(lr_prediction)
```

```python
svm_prediction=svr_rbf.predict(x_forecast)
print(svm_prediction)
```
